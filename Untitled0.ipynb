# 10. Verificação de Ano Bissexto
Escreva um programa que peça ao usuário um ano e determine se ele é bissexto ou não, exibindo a mensagem correspondente.

```python
# Programa para verificar se um ano é bissexto

ano = int(input("Digite um ano: "))

if (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0):
    print("O ano", ano, "é bissexto.")
else:
    print("O ano", ano, "não é bissexto.")
```

# 11. Cálculo de Multa por Excesso de Peso

João Papo-de-Pescador precisa controlar o redimento diário de seu trabalho. Desenvolva um programa que leia o peso de peixes. Caso o peso seja maior que 50 quilos, calcule o excesso e o valor da multa (R$ 4,00 por quilo excedente). Se náo houver excesso, os valores devem ser zero. Ao final, imprima o excesso e a multa

```python
peso = float(input("Digite o peso dos peixes (kg): "))

if peso > 50:
    excesso = peso - 50
    multa = excesso * 4
else:
    excesso = 0
    multa = 0

print("Excesso:", excesso, "kg")
print("Multa: R$", multa)
```

# 14. Loja de Tintas:

Faça um programa para uma loja de tintas. O programa deve pedir o tamanho em metros quadrados da área a ser pintada. Considere que a cobertura da tinta é de 1 litro para cada 3 metros quadrados e que a tinta é vendida em latas de 18 litros, que custam R$ 80,00. informe ao usuário a quantidade de latas de tinta a serem comprados e o preço total

```python
import math

area = float(input("Digite o tamanho da área a ser pintada (m²): "))

# quantidade de tinta necessária em litros
litros = area / 3

# quantidade de latas (arredondando para cima)
latas = math.ceil(litros / 18)

# preço total
preco = latas * 80

print("Quantidade de latas:", latas)
print("Preço total: R$", preco)
```

# 15. Validação de Nota:

Faça um programa que peça uma nota, entre zero e dez. Mostre uma mensagem caso o valor seja inválido e continue pedindo até que o usuário informe um valor válido.

```python
nota = float(input("Digite uma nota entre 0 e 10: "))

while nota < 0 or nota > 10:
    print("Valor inválido! Digite uma nota entre 0 e 10.")
    nota = float(input("Digite uma nota entre 0 e 10: "))

print("Nota válida:", nota)
```

# 16. Cadastro de Usuario e Senha:

Densenvolva um programa que leia um nome de usuario e a sua senha. O sistema não deve aceitar a senha igual ao nome do usuário, exibindo uma mensagem de erro e voltando a pedir as informações até que os dados sejam diferentes.

```python
import re

while True:
    usuario = input("Digite o nome de usuário: ")
    senha = input("Digite a senha: ")

    if senha == usuario:
        print("Erro: a senha não pode ser igual ao nome de usuário.")
        continue

    if len(senha) <= 8:
        print("Erro: a senha deve ter mais de 8 caracteres.")
        continue

    if not any(c.isupper() for c in senha):
        print("Erro: a senha deve ter pelo menos 1 letra maiúscula.")
        continue

    if not any(c.isdigit() for c in senha):
        print("Erro: a senha deve ter pelo menos 1 número.")
        continue

    if not re.search(r"[!@#$%^&*(),.?\":{}|<>]", senha):
        print("Erro: a senha deve ter pelo menos 1 símbolo.")
        continue

    print("Usuário e senha cadastrados com sucesso!")
    break
```

# 17. Gerador de Tabuada:

Desenvolva um gerador de tabuada, capaz de gerar a tabuada de qualquer número inteiro entre 1 a 10 conforme a escolha do usuário. A saida deve ser conforme o exemplo: "Tabuada do N: N x 1 = N" até o 10

```python
# Solicita ao usuário um número entre 1 e 10
n = int(input("Digite um número entre 1 e 10: "))

# Verifica se o número está no intervalo permitido
if 1 <= n <= 10:
    print(f"Tabuada do {n}:")

    for i in range(1, 11):
        print(f"{n} x {i} = {n * i}")
else:
    print("Por favor, digite um número entre 1 e 10.")
```

# 18. Média de Notas com Flag:

Crie um algoritimo que solicite ao usuário a digitaçáo de diversas notas. O programa deve calcular a média aritmética dessas notas. A entrada de dados deve ser interrompida quando o usuário digitar o valor 120 (condiçáo de parada), e este valor não deve ser contabilizado no cálculo da média.

```python
soma = 0
quantidade = 0

while True:
    nota = float(input("Digite uma nota (ou 120 para encerrar): "))

    if nota == 120:
        break

    soma = soma + nota
    quantidade = quantidade + 1

if quantidade > 0:
    media = soma / quantidade
    print("Média das notas:", media)
else:
    print("Nenhuma nota válida foi digitada.")
```

# 19. Maior e Menor em Vetor Aleatório:

Crie um progrma que gere um vetor com 10 números inteiros aleatorios entre 1 e 100. O algoritimo deve percorrer o vetor para indentificar e imprimir o maior e o menor valor encontrado, além de exibir o vetor completo

```python
import random

# Criando o vetor com 10 números aleatórios entre 1 e 100
vetor = []

for i in range(10):
    numero = random.randint(1, 100)
    vetor.append(numero)

# Inicializando maior e menor com o primeiro elemento
maior = vetor[0]
menor = vetor[0]

# Percorrendo o vetor para encontrar o maior e o menor valor
for num in vetor:
    if num > maior:
        maior = num
    if num < menor:
        menor = num

# Exibindo o vetor completo
print("Vetor gerado:", vetor)

# Exibindo o maior e o menor valor
print("Maior valor:", maior)
print("Menor valor:", menor)
```

# 20. Separação de Pares e ímpares:

Desenvolva um programa que preencha um vetor com 20 números inteiros aleatórios (entre 1 e 100). Em seguida, o programa deve separar esses números em dois novos vetores: um contendo apenas os números pares e outro apenas os números impares. Ao final, exiba os três vetores

```python
import random

# Vetores
vetor = []
pares = []
impares = []

# Preenchendo o vetor com 20 números aleatórios entre 1 e 100
for i in range(20):
    numero = random.randint(1, 100)
    vetor.append(numero)

# Separando números pares e ímpares
for num in vetor:
    if num % 2 == 0:
        pares.append(num)
    else:
        impares.append(num)

# Exibindo os vetores
print("Vetor original:", vetor)
print("Vetor de números pares:", pares)
print("Vetor de números ímpares:", impares)
```

# 21. Vetor de Padrão Alternado (0 e 1):

Escreva um algoritimo que preencha um vetor de 20 posições com um padrão alternado: as posições de indice par devem receber o valor 1 e as posições de índice impar devem receber o valor 0. Imprime o vetor resultante na tela

```python
# Criando o vetor
vetor = []

# Preenchendo o vetor com 20 posições
for i in range(20):
    if i % 2 == 0:   # índice par
        vetor.append(1)
    else:            # índice ímpar
        vetor.append(0)

# Imprimindo o vetor
print("Vetor resultante:", vetor)
```
