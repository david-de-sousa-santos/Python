# 10. Verificação de Ano Bissexto:
Escreva um programa que peça ao usuario um ano e determine se ele é bissexto ou não, exibindo a mensagem correspondente

```python
ano = int(input("Digite um ano: "))

if (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0):
      print(f"O ano {ano} é bissexto.")
else:
    print("O ano não é bissexto.")
```

# 11, Calculo de Multa por excesso de peso:
João Papo de pescador precisa controlar o rendimento diario de seu trabalho. Desenvolva um programa que leia o peso de peixes . Caso o peso seja maior que 50 quilos, Calcule o excesso e o valor da multa (R$ 4,00 por quilo excendente). Se não houver excesso, os valores devem sem ser zero. ao final,imprima o excesso e a multa .


```python
peso = float(input("Digite o peso dos peixes (kg): "))

limite = 50
multa_por_kg = 4

if peso > limite:
    excesso = peso - limite
    multa = excesso * multa_por_kg
else:
    excesso = 0
    multa = 0

print("Excesso de peso:", excesso, "kg")
print("Valor da multa: R$", multa)
```

# 13. Maior e menor de tres numeros faça o programa
Desenvolva um algoritimo que receba tres numeros inteiros e exiba, seoaradamente, qual e o maior e qual e o menir entre eles


```python
n1 = int(input("Digite o primeiro número: "))
n2 = int(input("Digite o segundo número: "))
n3 = int(input("Digite o terceiro número: "))

maior = n1
menor = n1

if n2 > maior:
    maior = n2
if n3 > maior:
    maior = n3

if n2 < menor:
    menor = n2
if n3 < menor:
    menor = n3

print("Maior número:", maior)
print("Menor número:", menor)
```

# 14. Loja de tintas
Faça um programa para um aloa de tintas. O programa de ve pedir o tamanho em metros quadrados a ser pintada . considere que a cobertura de tinta e de 1 litro para cada 3 metros quadrados e que a tinta é vendida em latas de 18 litros, que custam R$ 80,00. INFORME ao usuario a quatidade de latas de tin a serem compradas e o preço total

```python
import math

area = float(input("Digite o tamanho da área a ser pintada (m²): "))

litros_necessarios = area / 3
latas = math.ceil(litros_necessarios / 18)

preco = latas * 80

print("Quantidade de latas necessárias:", latas)
print("Preço total: R$", preco)
```

# 15. validação de nota
Faça um programa que peça uma nota , entre zero e dez . mostre uma mensagem caso o valos seja unvalido e continue pedindo ate que o usuario informe um valor valido

```python
nota = float(input("Digite uma nota entre 0 e 10: "))

while nota < 0 or nota > 10:
    print("Valor inválido! Digite uma nota entre 0 e 10.")
    nota = float(input("Digite a nota novamente: "))

print("Nota válida:", nota)

```

# 16. Cadastro de usuario e senha

```python
import re

usuario = input("Digite o nome de usuário: ")
senha = input("Digite a senha: ")

while (
    senha == usuario
    or len(senha) < 8
    or not re.search("[A-Z]", senha)
    or not re.search("[0-9]", senha)
    or not re.search("[!@#$%&*]", senha)
):
    print("Senha inválida!")
    print("A senha deve ter:")
    print("- Pelo menos 8 caracteres")
    print("- 1 letra maiúscula")
    print("- 1 número")
    print("- 1 símbolo (!@#$%&*)")
    print("- Não pode ser igual ao usuário")

    senha = input("Digite outra senha: ")

print("Cadastro realizado com sucesso!")
```

# 17. GERADOR DE TABUADA
Desenvolva um gerador de tabuada , capaz de gerar a tabuada de qualquer numero entre 1 a 10 conforme a escolha do usuario. A saida deve ser conforme o exemplo: " Tabuada do N: N x 1 = N" ATE O 10

```python
numero = int(input("Digite um número entre 1 e 10: "))

if numero >= 1 and numero <= 10:
    print("Tabuada do", numero)

    for i in range(1, 11):
        print(numero, "x", i, "=", numero * i)
else:
    print("Número inválido! Digite um valor entre 1 e 10.")
```
