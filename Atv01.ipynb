# 3. Calculadora de Segundos.
crie um programa que leia uma quantidade de dias, horas, minutos e segundos do usuario,calcule e mostre o total de segundos correspondente a esses valores


```python
# Calculadora de Segundos

dias = int(input("Dias: "))
horas = int(input("Horas: "))
minutos = int(input("Minutos: "))
segundos = int(input("Segundos: "))

total = (dias * 86400) + (horas * 3600) + (minutos * 60) + segundos

print("Total em segundos:", total)
```

# 4. Calculo de aumento salarial
Faça um programa que solicite o valor do salario atual e a porcentagem de aumento. O programa de ve calcular e imprimir o valor do aumento e o novo salario

```python
# 4. Cálculo de aumento salarial

salario = float(input("Salário atual: "))
percentual = float(input("Percentual de aumento (%): "))

aumento = salario * (percentual / 100)
novo_salario = salario + aumento

print("Valor do aumento:", aumento)
print("Novo salário:", novo_salario)
```

# 5. Calculo de desconto em mercadorias
Solicite o preço de uma mercadoria e o percentual de desconto. Calcule e exiba o valor do desconto e o preço final a pagar

```python
# 5. Cálculo de desconto em mercadorias

preco = float(input("Preço da mercadoria: "))
percentual = float(input("Percentual de desconto (%): "))

desconto = preco * (percentual / 100)
preco_final = preco - desconto

print("Valor do desconto:", desconto)
print("Preço final a pagar:", preco_final)
```

# 6. Tempo de viagem
escreva um programa que calcule o tempo estimado de uma viagem . O usuario deve informar a distancia em km e a velocidade média esperada , o Resultado deve ser exibido em horas e minutos

```python
# 6. Tempo de viagem

distancia = float(input("Distância (km): "))
velocidade = float(input("Velocidade média (km/h): "))

tempo_horas = distancia / velocidade

horas = int(tempo_horas)
minutos = int((tempo_horas - horas) * 60)

print(f"Tempo estimado: {horas} horas e {minutos} minutos")
```

# 7. Aluguel de carro
Desenvolva um algoritimo para calcular o preço do aluguel de um carro , o programa deve perguntar a quantidade de km percorridos e a quantidade de dias pelos quais o carrofoi alugado. calcule o preço total, sabendo que carro custa r 0,15 por km  rodado

```python
# 7. Aluguel de carro

km = float(input("Quilômetros percorridos: "))
dias = int(input("Quantidade de dias: "))

preco_km = 0.15
preco_dia = 60  # valor fixo por dia (opcional, pode ajustar se quiser)

total = (km * preco_km) + (dias * preco_dia)

print("Preço total do aluguel: R$", total)
```

# 8.Redução do tempo de vida de fumante.
crie um programa para calcular a reducão do tempo de vida de um fumante. pergunte a quantidade de cigarros fumados por dia e quantos anos ele ja fumou, considere que um fumante perde 10 min de vida a cada cigarro . calcule quantos dias de vida o fumante perdera a exiba o total


```python
# Programa para calcular a redução do tempo de vida de um fumante

# Entrada de dados
cigarros_por_dia = int(input("Quantos cigarros você fuma por dia? "))
anos_fumando = int(input("Há quantos anos você fuma? "))

# Cálculo
# Total de cigarros fumados
total_cigarros = cigarros_por_dia * 365 * anos_fumando

# Cada cigarro reduz 10 minutos de vida
total_minutos_perdidos = total_cigarros * 10

# Converter minutos para dias
dias_perdidos = total_minutos_perdidos / (60 * 24)

# Saída
print("Você perdeu aproximadamente", dias_perdidos, "dias de vida.")
```

# 9. Digitos de uma potencia grande
escreva um codigo que calcule a quantidade de digitos presentes no resultado da operacao de 2 elevado a 1.000.000

```python
import math

# Cálculo da quantidade de dígitos de 2^1.000.000
expoente = 1_000_000

# Fórmula: dígitos = ⌊ expoente * log10(2) ⌋ + 1
quantidade_digitos = int(expoente * math.log10(2)) + 1

print("Quantidade de dígitos em 2^1.000.000:", quantidade_digitos)
```
